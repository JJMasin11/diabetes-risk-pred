# Model Baseline
## 1.1 Import Data and Required Packages
### Importing Pandas, NumPy, MatPlotLib, Seaborn, and Warnings libraries

```python
# Basic Import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb
# Modeling
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.calibration import CalibratedClassifierCV
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier, early_stopping
import warnings
```

### Import the CSV data as Pandas dataframe

```python
df = pd.read_csv("../data/diabetes_012_health_indicators_BRFSS2015.csv")
```

### Show first 5 records

```python
df.head()
```

### Add BMI Age interaction

```python
df["BMI_Age"] = df["BMI"] * df["Age"]
```

### Preparing X and Y variables

```python
X = df.drop(columns=['Diabetes_012'], axis=1)
Y = df['Diabetes_012']
```

```python
# Create column transformer (Categorical features included in case some are engineered later)
num_features = X.select_dtypes(exclude="object").columns
cat_features = X.select_dtypes(include="object").columns

numeric_transformer = StandardScaler()
oh_transformer = OneHotEncoder()

preprocessor = ColumnTransformer(
    [
        ("OneHotEncoder", oh_transformer, cat_features),
        ("StandardScaler", numeric_transformer, num_features)
    ]
)
```

```python
X = preprocessor.fit_transform(X)
```

```python
# Train Test Split
x_train_full, x_test, y_train_full, y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)
x_train_full.shape, x_test.shape
```

```python
# Validation sets
x_train, x_val, y_train, y_val = train_test_split(x_train_full, y_train_full, test_size=0.2, random_state=42, stratify=y_train_full)
```

### Testing GPU

```python
# XGBoost
try:
    clf = XGBClassifier(tree_method="hist", device="cuda", n_estimators=1, max_depth=1)
    clf.fit(x_train[:100], y_train[:100]),
    print("XGBoost GPU test training complete!")
except xgb.core.XGBoostError as e:
    print("GPU training not available or failed:", e)
```

```python
# CatBoost
clf = CatBoostClassifier(task_type="GPU", iterations=1, depth=1, verbose=False)
clf.fit(x_train[:100], y_train[:100])
print("CatBoost GPU test training complete!")
```

```python
# LightGBM
clf = LGBMClassifier(device="gpu", n_estimators=1, max_depth=1)
clf.fit(x_train[:100], y_train[:100])
print("LightGBM GPU test training complete!")
```

### Create an evaluation function to give all metrics after training

```python
def evaluate_model(true, predicted, proba):
    accuracy = accuracy_score(true, predicted)
    precision = precision_score(true, predicted, average="macro", zero_division=0)
    recall = recall_score(true, predicted, average="macro")
    f1 = f1_score(true, predicted, average="macro")
    roc_auc = roc_auc_score(true, proba, average="macro", multi_class="ovr")
    matrix = confusion_matrix(true, predicted)
    
    return accuracy, precision, recall, f1, roc_auc, matrix
```

### Baseline

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest Classifier": RandomForestClassifier(),
    "Support Vector Machine": CalibratedClassifierCV(LinearSVC(), method="sigmoid"),
    "K-Neighbors Classifier": KNeighborsClassifier(n_jobs=-1),
    "XGBoost Classifier": XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, subsample=0.8, colsample_bytree=0.8, tree_method="hist", device="cuda", n_jobs=-1, eval_metric="mlogloss", objective="multi:softprob", early_stopping_rounds=10),
    "CatBoost Classifier": CatBoostClassifier(n_estimators=500, depth=6, learning_rate=0.1, task_type="GPU", verbose=False, early_stopping_rounds=50, loss_function="MultiClass"),
    "LightGBM Classifier": LGBMClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, subsample=0.8, colsample_bytree=0.8, device="gpu", verbose=0, objective="multiclass", num_class=3)
}
model_list = []
rec_list = []
roc_auc_list = []
f1_list = []
combined_list = []
confusion_matrix_list = []

for i in range(len(list(models))):

    # Train model
    model = list(models.values())[i]

    if model.__class__.__name__ == "LGBMClassifier":
        callbacks = [early_stopping(stopping_rounds=10, verbose=0)]
        model.fit(x_train, y_train, eval_set=[(x_val, y_val)], eval_metric="multi_logloss", callbacks=callbacks)
    elif model.__class__.__name__ == "XGBClassifier":
        model.fit(x_train, y_train, eval_set=[(x_val, y_val)], verbose=0)
    elif model.__class__.__name__ == "CatBoostClassifier":
        model.fit(x_train, y_train, eval_set=(x_val, y_val))
    else:
        model.fit(x_train_full, y_train_full)

    # Make predictions
    y_train_pred = model.predict(x_train_full)
    y_test_pred = model.predict(x_test)

    # Get predicted probabilities for ROC AUC calculation
    y_train_proba = model.predict_proba(x_train_full)
    y_test_proba = model.predict_proba(x_test)
    print(y_train_proba.shape)

    # Evaluate model
    train_acc, train_prec, train_rec, train_f1, train_roc_auc, train_matrix = evaluate_model(y_train_full, y_train_pred, y_train_proba)
    test_acc, test_prec, test_rec, test_f1, test_roc_auc, test_matrix = evaluate_model(y_test, y_test_pred, y_test_proba)

    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])

    print("Model performance for training set:")
    print("- Accuracy: {:.4f}".format(train_acc))
    print("- Precision: {:.4f}".format(train_prec))
    print("- Recall: {:.4f}".format(train_rec))
    print("- F1 Score: {:.4f}".format(train_f1))
    print("- ROC AUC: {:.4f}".format(train_roc_auc))

    print("-------------------------------------------")

    print("Model performance for test set:")
    print("- Accuracy: {:.4f}".format(test_acc))
    print("- Precision: {:.4f}".format(test_prec))
    print("- Recall: {:.4f}".format(test_rec))
    print("- F1 Score: {:.4f}".format(test_f1))
    print("- ROC AUC: {:.4f}".format(test_roc_auc))

    rec_list.append(test_rec)
    roc_auc_list.append(test_roc_auc)
    f1_list.append(test_f1)

    # Combined metric (weighted average of recall, roc_auc, and f1 score)
    combined_score = (0.5 * test_rec) + (0.3 * test_roc_auc) + (0.2 * test_f1)
    combined_list.append(combined_score)

    confusion_matrix_list.append(test_matrix)

    print('=' * 35)
    print("\n")
```

### Results

```python
results = pd.DataFrame(list(zip(model_list, rec_list, roc_auc_list, f1_list, combined_list)), columns=["Model", "Recall", "ROC-AUC", "F1 Score", "Combined Score"]).sort_values(by="Combined Score", ascending=False)
```

```python
# Save baseline to CSV
results.to_csv("baseline_results.csv", index=False)
```
