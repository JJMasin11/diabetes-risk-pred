# Hyperparameter Tuning
# 1.1 Import Data and Required Packages
### Importing Pandas, NumPy, MatPlotLib, Seaborn, and Warnings libraries

```python
# Basic Import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Modeling
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, precision_recall_curve
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier, early_stopping
import warnings
```

### Import the CSV data as Pandas dataframe

```python
df = pd.read_csv("../data/diabetes_012_health_indicators_BRFSS2015.csv")
```

### Add BMI Age interaction

```python
df["BMI_Age"] = df["BMI"] * df["Age"]
```

### Preparing X and Y variables

```python
X = df.drop(columns=["Diabetes_012"])
Y = df["Diabetes_012"]
y_binary = (Y != 0).astype(int)
```

```python
# Create column transformer (Categorical features included in case some are engineered later)
num_features = X.select_dtypes(exclude=["object"]).columns
cat_features = X.select_dtypes(include=["object"]).columns

numeric_transformer = StandardScaler()
oh_transformer = OneHotEncoder()

preprocessor = ColumnTransformer(
    [
        ("OneHotEncoder", oh_transformer, cat_features),
        ("StandardScaler", numeric_transformer, num_features),
    ]
)
```

```python
X = preprocessor.fit_transform(X)
```

```python
# Train-test split
x_train, x_test, y_train, y_test = train_test_split(X, y_binary, test_size=0.2, random_state=42, stratify=y_binary)
x_train.shape, x_test.shape
```

```python
# Validation sets
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42, stratify=y_train)
```

### Create an evaluation function to give all metrics after training

```python
def evaluate_model(true, predicted, proba):
    accuracy = accuracy_score(true, predicted)
    precision = precision_score(true, predicted, zero_division=0)
    recall = recall_score(true, predicted)
    f1 = f1_score(true, predicted)
    roc_auc = roc_auc_score(true, proba)
    conf_matrix = confusion_matrix(true, predicted)
    pr_curve = precision_recall_curve(true, proba)

    return accuracy, precision, recall, f1, roc_auc, conf_matrix, pr_curve
```

```python
# Get scale_pos_weight
neg, pos = np.bincount(y_train)

models = {
    "XGBoost Classifier": XGBClassifier(tree_method="hist", device="cuda", n_jobs=-1, eval_metric="logloss", objective="binary:logistic", scale_pos_weight=neg / pos),
    "CatBoost Classifier": CatBoostClassifier(task_type="GPU", verbose=False, loss_function="Logloss", auto_class_weights="Balanced"),
    "LightGBM Classifier": LGBMClassifier(device="gpu", verbose=0, objective="binary", class_weight="balanced"),
}
model_list = []
rec_list = []
prec_list = []
roc_auc_list = []
f1_list = []
confusion_matrix_list = []
pr_curve_list = []

# Parameter Distributions
xgb_param_dist = {
    "max_depth": [3, 5, 7],
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [200, 500, 1000],
    "subsample": [0.7, 0.9, 1.0],
    "colsample_bytree": [0.7, 0.9, 1.0]
}

cat_param_dist = {
    "depth": [4, 6, 8],
    "learning_rate": [0.01, 0.05, 0.1],
    "iterations": [300, 600, 1000]
}

lgbm_param_dist = {
    "num_leaves": [31, 50, 100],
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [200, 500, 1000],
    "subsample": [0.7, 0.9, 1.0]
}

for i in range(len(list(models))):

    # Train model
    model = list(models.values())[i]

    # Call RandomizedSearchCV with appropriate parameter distributions and fit methods
    if model.__class__.__name__ == "LGBMClassifier":
        random_search = RandomizedSearchCV(model, lgbm_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train, y_train, eval_set=[(x_val, y_val)], eval_metric="logloss")
    elif model.__class__.__name__ == "XGBClassifier":
        random_search = RandomizedSearchCV(model, xgb_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train, y_train, eval_set=[(x_val, y_val)], verbose=0)
    else:
        random_search = RandomizedSearchCV(model, cat_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train, y_train, eval_set=(x_val, y_val))

    # Get predicted probabilities
    y_train_proba = random_search.predict_proba(x_train)
    y_test_proba = random_search.predict_proba(x_test)

    y_train_proba_pos = y_train_proba[:, 1]
    y_test_proba_pos = y_test_proba[:, 1]

    # Make predictions
    y_train_pred = random_search.predict(x_train)
    y_test_pred = random_search.predict(x_test)

    # Evaluate model
    train_acc, train_prec, train_rec, train_f1, train_roc_auc, train_conf_matrix, train_pr_curve = evaluate_model(y_train, y_train_pred, y_train_proba_pos)
    test_acc, test_prec, test_rec, test_f1, test_roc_auc, test_conf_matrix, test_pr_curve = evaluate_model(y_test, y_test_pred, y_test_proba_pos)

    print(list(models.keys())[i])
    print("Best Hyperparameters:", random_search.best_params_)

    model_list.append(list(models.keys())[i])
    print("Model performance for training set:")
    print("- Accuracy: {:.4f}".format(train_acc))
    print("- Precision: {:.4f}".format(train_prec))
    print("- Recall: {:.4f}".format(train_rec))
    print("- F1 Score: {:.4f}".format(train_f1))
    print("- ROC AUC: {:.4f}".format(train_roc_auc))

    print("-------------------------------------------")

    print("Model performance for test set:")
    print("- Accuracy: {:.4f}".format(test_acc))
    print("- Precision: {:.4f}".format(test_prec))
    print("- Recall: {:.4f}".format(test_rec))
    print("- F1 Score: {:.4f}".format(test_f1))
    print("- ROC AUC: {:.4f}".format(test_roc_auc))

    rec_list.append(test_rec)
    prec_list.append(test_prec)
    roc_auc_list.append(test_roc_auc)
    f1_list.append(test_f1)

    confusion_matrix_list.append(test_conf_matrix)    
    pr_curve_list.append(test_pr_curve)

    print('=' * 35)
    print("\n")
```

Randomized search found that the best hyperparameters for each model are as follows:
* XGBoost: max_depth=5, learning_rate=0.01, n_estimators=1000, subsample=0.7, colsample_bytree=0.7
* CatBoost: depth=4, learning_rate=0.05, iterations=600
* LightGBM: num_leaves=31, learning_rate=0.01, n_estimators=1000, subsample=0.7

### Threshold Testing on tuned hyperparameters

```python
neg, pos = np.bincount(y_train)

tuned_models = {
    "XGBoost Classifier": XGBClassifier(max_depth=5, learning_rate=0.01, n_estimators=1000, subsample=0.7, colsample_bytree=0.7, tree_method="hist", device="cuda", n_jobs=-1, eval_metric="logloss", objective="binary:logistic", scale_pos_weight=(neg / pos)),
    "CatBoost Classifier": CatBoostClassifier(depth=4, learning_rate=0.05, iterations=600, task_type="GPU", verbose=False, loss_function="Logloss", scale_pos_weight=(neg / pos)),
    "LightGBM Classifier": LGBMClassifier(num_leaves=31, learning_rate=0.01, n_estimators=1000, subsample=0.7, device="gpu", verbose=0, objective="binary", scale_pos_weight=(neg / pos))
}
model_list = []
rec_list = []
prec_list = []
roc_auc_list = []
f1_list = []
confusion_matrix_list = []
pr_curve_list = []

for i in range(len(list(tuned_models))):

    # Train model
    model = list(tuned_models.values())[i]

    # Call correct fit method
    if model.__class__.__name__ == "LGBMClassifier":
        model.fit(x_train, y_train, eval_set=[(x_val, y_val)], eval_metric="logloss")
    elif model.__class__.__name__ == "XGBClassifier":
        model.fit(x_train, y_train, eval_set=[(x_val, y_val)], verbose=0)
    else:
        model.fit(x_train, y_train, eval_set=(x_val, y_val))

    # Get predicted probabilities
    y_train_proba = model.predict_proba(x_train)
    y_test_proba = model.predict_proba(x_test)

    y_train_proba_pos = y_train_proba[:, 1]
    y_test_proba_pos = y_test_proba[:, 1]

    # Make predictions
    y_train_pred = model.predict(x_train)
    y_test_pred = model.predict(x_test)

    # Evaluate model
    train_acc, train_prec, train_rec, train_f1, train_roc_auc, train_conf_matrix, train_pr_curve = evaluate_model(y_train, y_train_pred, y_train_proba_pos)
    test_acc, test_prec, test_rec, test_f1, test_roc_auc, test_conf_matrix, test_pr_curve = evaluate_model(y_test, y_test_pred, y_test_proba_pos)

    print(list(models.keys())[i])

    model_list.append(list(models.keys())[i])
    print("Model performance for training set:")
    print("- Accuracy: {:.4f}".format(train_acc))
    print("- Precision: {:.4f}".format(train_prec))
    print("- Recall: {:.4f}".format(train_rec))
    print("- F1 Score: {:.4f}".format(train_f1))
    print("- ROC AUC: {:.4f}".format(train_roc_auc))

    print("-------------------------------------------")

    print("Model performance for test set:")
    print("- Accuracy: {:.4f}".format(test_acc))
    print("- Precision: {:.4f}".format(test_prec))
    print("- Recall: {:.4f}".format(test_rec))
    print("- F1 Score: {:.4f}".format(test_f1))
    print("- ROC AUC: {:.4f}".format(test_roc_auc))

    rec_list.append(test_rec)
    prec_list.append(test_prec)
    roc_auc_list.append(test_roc_auc)
    f1_list.append(test_f1)

    confusion_matrix_list.append(test_conf_matrix)    
    pr_curve_list.append(test_pr_curve)

    print('=' * 35)
    print("\n")
```

```python
# Find best threshold for each model to maximize precision while maintaining recall above 0.8
target_recall = 0.8

for model in range(len(model_list)):
    best_threshold = 0
    best_precision = 0

    precision, recall, thresholds = pr_curve_list[model]

    for t in range(len(thresholds)):
        if recall[t] >= target_recall:
            if precision[t] > best_precision:
                best_precision = precision[t]
                best_threshold = thresholds[t]

    print("Best threshold for {}: {:.2f} with precision {:.4f} and recall {:.4f}".format(model_list[model], best_threshold, best_precision, target_recall))
```

### Results

```python
results = pd.DataFrame(list(zip(model_list, rec_list, prec_list, roc_auc_list, f1_list)), columns=["Model", "Recall", "Precision", "ROC AUC", "F1 Score"]).sort_values(by="Recall", ascending=False)
```
