# Diabetes Risk Prediction Using Machine Learning
**Binary Classification: No Diabetes vs. At-Risk**

## Executive Summary

This project developed a machine learning pipeline for predicting diabetes risk using healthcare survey data. Several machine learning models were evaluated, including Logistic Regression, XGBoost, LightGBM, and CatBoost, with gradient boosting approaches producing the strongest overall performance.

After hyperparameter tuning and threshold optimization, the final CatBoost model achieved a recall of 0.80, precision of 0.33, and ROC-AUC of 0.825. Threshold optimization was used to align model behavior with the healthcare-oriented objective of maximizing identification of at-risk individuals while minimizing false negatives.

To improve robustness during inference, an auxiliary CatBoost cholesterol prediction model was developed to handle uncertain user inputs through probabilistic feature imputation. The final system demonstrates a production-oriented machine learning workflow incorporating preprocessing, imbalance handling, threshold optimization, probabilistic inference, and deployment-ready model persistence.

## 1. Problem Statement
The objective of this project is to predict whether a patient is at risk of diabetes using clinical and demographic features. The original dataset contained three classes: 0 (no diabetes), 1 (prediabetes), and 2 (diabetes), Due to significant feature overlap and limited separability between the prediabetes and diabetes classes, these were merged into a single "at-risk" class, transforming the problem into a binary classification task.

Given the healthcare context, the primary objective is to maximize recall in order to minimize the false negatives and ensure that as many at-risk patients as possible are correctly identified. Precision was also monitored to balance the rate of false positives.

## 4. Data Preprocessing Pipeline

````python
X = df.drop(columns=["Diabetes_Binary"], axis=1)
Y = df["Diabetes_Binary"]
````

````python
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer

# Create column transformer
num_features = X.select_dtypes(exclude=["object"]).columns

numeric_transformer = StandardScaler()

preprocessor = ColumnTransformer(
    [
        ("StandardScaler", numeric_transformer, num_features)
    ]
)
````

````python
X = preprocessor.fit_transform(X)
````

````python
# Train test split
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)
````

The features were preprocessed using `StandardScalar` to standardize the data by centering each feature around zero mean and unit variance. Although scaling is not strictly required for tree-based models, it was applied to maintain consistency across the preprocessing pipeline and ensure compatibility with a wider range of models during experimentation.

The dataset was split into training and testing sets using a stratified `train_test_split` to preserve the original class distribution in both subsets. This is particularly important given the class imbalance in the dataset, as it ensures that both the training and testing sets contain representative proportions of at-risk and non-risk patients.

## 5. Handling Class Imbalance

Due to significant class imbalance in the dataset, imbalance handling techniques were applied to reduce bias toward the majority class and improve recall for at-risk patients. Without adjustment, models trained on imbalanced data may disproportionately favor the majority class, leading to poor identification of minority-class samples.

For the scikit-learn models, `class_weight='balanced'` was used to automatically assign higher weights to the minority class during training. This increases the penalty for misclassifying at-risk patients and helps the models better account for the imbalanced class distribution.

For the gradient boosting models (CatBoost, XGBoost, and LightGBM), `scale_pos_weight = neg / pos` was used to increase the contribution of positive-class errors within the loss function. This encourages the models to place greater emphasis on correctly identifying at-risk patients during training.

## 6. Baseline Model

A Logistic Regression model was used to establish baseline performance for the binary classification task. Logistic Regression was selected due to its simplicity, interpretability, and effectiveness as a benchmark for evaluating more complex non-linear models.

````python
from sklearn.linear_model import LogisticRegression

baseline_model = LogisticRegression(random_state=42)
baseline_model.fit(x_train, y_train)

# Make predictions
y_pred_base = baseline_model.predict(x_test)

# Get predicted probabilities
y_proba_base = baseline_model.predict_proba(x_test)[:, 1]

from sklearn.metrics import precision_score, recall_score, roc_auc_score

precision_base = precision_score(y_test, y_pred_base)
recall_base = recall_score(y_test, y_pred_base)
roc_auc_base = roc_auc_score(y_test, y_proba_base)

print("Precision:", precision_base)
print("Recall:", recall_base)
print("ROC AUC:", roc_auc_base)
````

The baseline model achieved the following performance metrics:
* Precision: 0.562
* Recall: 0.199
* ROC-AUC: 0.819

The baseline evaluation was initially performed using the default classification threshold of 0.5.

While the model achieved relatively high precision and strong ROC-AUC performance, recall was significantly lower than desired for the healthcare-focused objective of the project. This indicates that the model was overly conservative in predicting the at-risk class and failed to identify a large proportion of at-risk patients.

These results align with the exploratory data analysis, which showed substantial feature overlap and weak linear separability between classes. As a result, more flexible non-linear ensemble methods such as XGBoost, LightGBM, and CatBoost were explored to better capture feature interactions and improve recall performance.

## 7. Advanced Model Training

Following the baseline Logistic Regression model, several gradient boosting ensemble methods were evaluated to better capture the complex non-linear relationships present in the dataset. Exploratory data analysis showed substantial feature overlap and weak linear separability between classes, suggesting that more flexible models would be required to improve minority-class recall.

The advanced models explored in this project included XGBoost, LightGBM, and CatBoost. These models were selected due to their strong performance on structured tabular datasets and their ability to model non-linear feature interactions effectively.

To address the class imbalance in the dataset, class weighting techniques were incorporated during training. The models generated probability estimates that were used for ROC-AUC evaluation and later threshold optimization to balance recall and precision according to the project objectives.

````python
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
import numpy as np

neg, pos = np.bincount(y_train)

models = {
    "XGBoost Classifier": XGBClassifier(eval_metric="logloss", objective="binary:logistic", tree_method="hist", scale_pos_weight=neg / pos, device="cuda", n_jobs=-1, random_state=42, verbosity=0),
    "CatBoost Classifier": CatBoostClassifier(loss_function="Logloss", auto_class_weights="Balanced", task_type="GPU", random_state=42, verbose=0),
    "LightGBM Classifier": LGBMClassifier(objective="binary", class_weight="balanced", device="gpu", random_state=42, verbosity=-1)
}
model_list = []
prec_list = []
rec_list = []
roc_auc_list = []

for i in range(len(list(models))):

    # Train model
    model = list(models.values())[i]

    # Call correct fit method
    if model.__class__.__name__ == "LGBMClassifier":
        model.fit(x_train, y_train, eval_metric="logloss")
    else:
        model.fit(x_train, y_train)

    # Make predictions
    y_pred = model.predict(x_test)

    # Get predicted probabilities
    y_proba = model.predict_proba(x_test)[:, 1]

    # Calculate metrics
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_proba)

    model_list.append(list(models.keys())[i])
    prec_list.append(precision)
    rec_list.append(recall)
    roc_auc_list.append(roc_auc)

    print(f"{list(models.keys())[i]} performance:")
    print("- Precision: {:.4f}".format(precision))
    print("- Recall: {:.4f}".format(recall))
    print("- ROC AUC: {:.4f}".format(roc_auc))

    print('=' * 35)
    print("\n")
````

Initial evaluations showed that the gradient boosting models substantially improved recall compared to the Logistic Regression baseline while maintaining strong ROC-AUC performance. However, precision remained relatively low, indicating that additional tuning and threshold optimization would be necessary to better balance false positives and false negatives prior to final model selection.

A table of the initial evaluation metrics can be seen below:

|           | XGBoost | CatBoost | LightGBM |
|-----------|---------|----------|----------|
| Precision | 0.338   | 0.337    | 0.335    |
| Recall    | 0.766   | 0.785    | 0.785    |
| ROC-AUC   | 0.818   | 0.825    | 0.825    |

## 8. Hyperparameter Tuning

Hyperparameter tuning was performed using `RandomizedSearchCV` to improve model generalization performance while effectively exploring the search space. Randomized search was selected instead of exhaustive grid search due to its lower computational cost and ability to evaluate a broader range of parameter combinations within a limited computational budget.

The selected parameter distributions focused primarily on controlling:
* Model complexity (`max_depth`, `depth`, `num_leaves`)
* Learning stability (`learning_rate`)
* Ensemble strength (`n_estimators`, `iterations`)
* Overfitting reduction through stochastic sampling (`subsample`, `colsample_bytree`)

The parameter ranges were intentionally constrained to moderate values in order to balance model flexibility, training efficiency, and generalization performance while reducing the risk of overfitting on the dataset.

ROC-AUC was used as the primary scoring metric during hyperparameter tuning because it evaluates the model's ability to distinguish between positive and negative classes across all possible classification thresholds. Unlike recall alone, ROC-AUC provides a threshold-independent measure of ranking performance and helps prevent models from artificially maximizing recall by excessively predicting the positive class.

Since the final project workflow included later threshold optimization, ROC-AUC provided a strong objective for identifying models with robust probability-ranking performance prior to selecting an operational decision threshold.

````python
from sklearn.model_selection import RandomizedSearchCV

# Get validation sets
x_train_param, x_val, y_train_param, y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42, stratify=y_train)

neg, pos = np.bincount(y_train)

models = {
    "XGBoost Classifier": XGBClassifier(eval_metric="logloss", objective="binary:logistic", tree_method="hist", scale_pos_weight=neg / pos, device="cuda", n_jobs=-1, random_state=42, verbosity=0),
    "CatBoost Classifier": CatBoostClassifier(loss_function="Logloss", auto_class_weights="Balanced", task_type="GPU", random_state=42, verbose=0),
    "LightGBM Classifier": LGBMClassifier(objective="binary", class_weight="balanced", device="gpu", random_state=42, verbosity=-1)
}
model_list = []
prec_list = []
rec_list = []
roc_auc_list = []

# Parameter Distributions
xgb_param_dist = {
    "max_depth": [3, 5, 7],
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [200, 500, 1000],
    "subsample": [0.7, 0.9, 1.0],
    "colsample_bytree": [0.7, 0.9, 1.0]
}

cat_param_dist = {
    "depth": [4, 6, 8],
    "learning_rate": [0.01, 0.05, 0.1],
    "iterations": [300, 600, 1000]
}

lgbm_param_dist = {
    "num_leaves": [31, 50, 100],
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators": [200, 500, 1000],
    "subsample": [0.7, 0.9, 1.0]
}

for i in range(len(list(models))):

    # Train model
    model = list(models.values())[i]

    # Call RandomizedSearchCV with appropriate parameter distributions and fit methods
    if model.__class__.__name__ == "LGBMClassifier":
        random_search = RandomizedSearchCV(model, lgbm_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train_param, y_train_param, eval_set=[(x_val, y_val)], eval_metric="logloss")
    elif model.__class__.__name__ == "XGBClassifier":
        random_search = RandomizedSearchCV(model, xgb_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train_param, y_train_param, eval_set=[(x_val, y_val)])
    else:
        random_search = RandomizedSearchCV(model, cat_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
        random_search.fit(x_train_param, y_train_param, eval_set=[(x_val, y_val)])

    # Make predictions
    y_pred = random_search.predict(x_test)

    # Get predicted probabilities
    y_proba = random_search.predict_proba(x_test)[:, 1]

    # Calculate metrics
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_proba)

    model_list.append(list(models.keys())[i])
    prec_list.append(precision)
    rec_list.append(recall)
    roc_auc_list.append(roc_auc)

    print(f"{list(models.keys())[i]} performance after hyperparameter tuning:")
    print("Best Hyperparameters:", random_search.best_params_)
    print("- Precision: {:.4f}".format(precision))
    print("- Recall: {:.4f}".format(recall))
    print("- ROC AUC: {:.4f}".format(roc_auc))

    print('=' * 35)
    print("\n")
````

The table below shows the best hyperparameters for each of the models:

| Model | Best Hyperparameters |
|-------|----------------------|
| XGBoost | `learning_rate=0.01, n_estimators=1000, max_depth=5, subsample=0.7, colsample_bytree=0.7`|
| CatBoost | `learning_rate=0.01, iterations=1000, depth=6` |
| LightGBM | `learning_rate=0.01, n_estimators=1000, num_leaves=31, subsample=0.7` |

The tuned configurations consistently favored smaller learning rates paired with larger boosting iteration counts, indicating that gradual learning produced more stable generalization performance on the dataset. Moderate tree complexity and subsampling parameters further suggest that controlling overfitting was important given the significant feature overlap between classes.

## 9. Threshold Optimization

Threshold optimization was performed using precision, recall, and threshold values generated by `precision_recall_curve`. Rather than relying on the default decision threshold of 0.5, thresholds were selected to maximize precision while maintaining recall above 0.80.

A recall constraint of 0.80 was chosen due to the healthcare-focused nature of the problem, where minimizing false negatives is especially important in order to identify as many at-risk patients as possible. Because increasing precision typically reduces recall, threshold selection involved balancing the tradeoff between false positives and false negatives.

The tuned model consistently produced strong ROC-AUC scores, indicating effective probability ranking performance across all three gradient boosting approaches. As a result, threshold selection became the primary mechanism for controlling the operational tradeoff between false positives and false negatives.

````python
from sklearn.metrics import precision_recall_curve

neg, pos = np.bincount(y_train)
target_recall = 0.8

tuned_models = {
    "XGBoost Classifier": XGBClassifier(learning_rate=0.01, n_estimators=1000, max_depth=5, subsample=0.7, colsample_bytree=0.7, eval_metric="logloss", objective="binary:logistic", tree_method="hist", scale_pos_weight=neg / pos, device="cuda", n_jobs=-1, random_state=42, verbosity=0),
    "CatBoost Classifier": CatBoostClassifier(learning_rate=0.01, iterations=1000, depth=6, loss_function="Logloss", auto_class_weights="Balanced", task_type="GPU", random_state=42, verbose=0),
    "LightGBM Classifier": LGBMClassifier(learning_rate=0.01, n_estimators=1000, num_leaves=31, subsample=0.7, objective="binary", class_weight="balanced", device="gpu", random_state=42, verbosity=-1)
}
model_list = []
thresh_list = []
prec_list = []
rec_list = []
roc_auc_list = []

for i in range(len(list(tuned_models))):
    
    # Train model
    tuned_model = list(tuned_models.values())[i]

    # Call correct fit method
    if tuned_model.__class__.__name__ == "LGBMClassifier":
        tuned_model.fit(x_train, y_train, eval_metric="logloss")
    else:
        tuned_model.fit(x_train, y_train)

    # Get predicted probabilities
    y_proba_tuned = tuned_model.predict_proba(x_test)[:, 1]

    # Find best threshold to maximize precision while maintaining recall above 0.8
    best_threshold = 0
    best_precision = 0

    precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba_tuned)

    for t in range(len(thresholds)):
        if recalls[t] >= target_recall and precisions[t] > best_precision:
            best_precision = precisions[t]
            best_threshold = thresholds[t]

    thresh_list.append(best_threshold)

    # Make predictions using best threshold
    y_pred_tuned = (y_proba_tuned >= best_threshold).astype(int)

    # Calculate metrics
    precision = precision_score(y_test, y_pred_tuned)
    recall = recall_score(y_test, y_pred_tuned)
    roc_auc = roc_auc_score(y_test, y_proba_tuned)

    model_list.append(list(tuned_models.keys())[i])
    prec_list.append(precision)
    rec_list.append(recall)
    roc_auc_list.append(roc_auc)

    print(f"{list(tuned_models.keys())[i]} performance after threshold tuning:")
    print("Best Threshold:", best_threshold)
    print("- Precision: {:.4f}".format(precision))
    print("- Recall: {:.4f}".format(recall))
    print("- ROC AUC: {:.4f}".format(roc_auc))

    print('=' * 35)
    print("\n")
````

The optimal operating threshold for all three models converged near 0.48, producing the highest precision values while maintaining the target recall requirement of 0.80.

The table below shows the precision and recall achieved at the selected operating threshold for each model, along with the corresponding ROC-AUC scores:

| Model | Threshold | Precision | Recall | ROC-AUC |
|-------|-----------|-----------|--------|---------|
| XGBoost | 0.48 | 0.328 | 0.801 | 0.825 |
| CatBoost | 0.48 | 0.329 | 0.800 | 0.825 |
| LightGBM | 0.48 | 0.328 | 0.801 | 0.825 |

## 10. Final Model Selection

````python
from catboost import CatBoostClassifier
from sklearn.metrics import precision_score, recall_score, roc_auc_score

threshold = 0.48

final_model = CatBoostClassifier(learning_rate=0.01, iterations=1000, depth=6, loss_function="Logloss", auto_class_weights="Balanced", task_type="GPU", random_state=42, verbose=0)
final_model.fit(x_train, y_train)

# Get predicted probabilities
y_proba_final = final_model.predict_proba(x_test)[:, 1]

# Make predictions using best threshold
y_pred_final = (y_proba_final >= threshold).astype(int)

# Calculate metrics
precision_final = precision_score(y_test, y_pred_final)
recall_final = recall_score(y_test, y_pred_final)
roc_auc_final = roc_auc_score(y_test, y_proba_final)

print("Final Model Performance:")
print("- Precision: {:.4f}".format(precision_final))
print("- Recall: {:.4f}".format(recall_final))
print("- ROC AUC: {:.4f}".format(roc_auc_final))
````

Although XGBoost, LightGBM, and CatBoost achieved nearly identical performance after hyperparameter tuning and threshold optimization, CatBoost was selected for deployment due to its robustness, consistent probability behavior during threshold optimization, and reduced sensitivity to feature preprocessing. Given the performance convergence across the models, deployment reliability and operational stability were prioritized over marginal metric differences.

The final CatBoost model achieved approximately 0.80 recall while maintaining precision near 0.33 at the selected operating threshold, aligning with the project objective of maximizing identification of at-risk patients while limiting false positives as effectively as possible.

The final deployed system therefore consists of the trained CatBoost model paired with the optimized operating threshold identified during threshold optimization.

## 12. Handling Uncertain User Inputs

### Motivation

Since many users may not know their cholesterol values or may have never had their cholesterol tested, the frontend was designed to include an "unsure" option for cholesterol status inputs. However, selecting this option introduces missing data that must be handled before inference can be performed with the primary CatBoost diabetes prediction model. While naive approaches such as mode imputation are computationally simple, they may reduce overall prediction quality by failing to account for relationships between cholesterol status and other health indicators. To address this, an auxiliary cholesterol prediction model was developed and evaluated as an alternative imputation strategy.

### Synthetic Missingness Simulation

Because the dataset did not contain naturally missing cholesterol values, synthetic missingness was introduced to evaluate imputation performance under realistic inference conditions. Cholesterol values were artificially masked in the test dataset to emulate users selecting the "unsure" option during inference.

This was achieved by creating a mask using `np.random.rand(len(chol_test_df))` which generates a NumPy array of uniformly distributed random numbers between 0 and 1 with the same length as the test set. The array is then converted into Boolean values by evaluating whether each number is less than 0.2, resulting in approximately 20% of rows being marked as true. The mask is then used to selectively replace the HighChol values in those rows with NaN using `test_missing.loc[mask, "HighChol"] = np.nan`. This synthetically introduces missing cholesterol values to simulate users selecting an "unsure" option during inference.

````python
# Drop Diabetes_Binary since it won't be known at prediction time
chol_df = df.drop(columns=["Diabetes_Binary"])

# Split into train and test sets
chol_train_df, chol_test_df = train_test_split(chol_df, test_size=0.2, random_state=42, stratify=df["HighChol"])
y_chol_true = chol_test_df["HighChol"].copy()

# Simulate missing values
mask = np.random.rand(len(chol_test_df)) < 0.2
test_missing = chol_test_df.copy()
test_missing.loc[mask, "HighChol"] = np.nan
````

### Baseline Mode Imputation

As a baseline comparison, mode imputation was performed on the synthetically masked test set. Missing HighChol values were replaced using the most common cholesterol class observed in the training dataset. This provided a simple statistical imputation strategy against which the auxiliary cholesterol prediction model could be evaluated.

````python
mode = chol_train_df["HighChol"].mode()[0]
test_missing["HighChol"] = test_missing["HighChol"].fillna(mode)

y_chol_eval_true = y_chol_true[mask]
y_chol_eval_pred = test_missing.loc[mask, "HighChol"]

positive_rate = chol_train_df["HighChol"].mean()
y_chol_eval_proba = np.full(len(y_chol_eval_true), positive_rate)

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_chol_eval_true, y_chol_eval_pred)
precision = precision_score(y_chol_eval_true, y_chol_eval_pred)
recall = recall_score(y_chol_eval_true, y_chol_eval_pred)
roc_auc = roc_auc_score(y_chol_eval_true, y_chol_eval_proba)

print("Mode Imputation Performance:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"ROC AUC: {roc_auc:.4f}")
````

| Metric | Value |
|--------|-------|
| Accuracy | 0.579 |
| Precision | 0.0 |
| Recall | 0.0 |
| ROC-AUC | 0.5 |

Although mode imputation achieved an accuracy of 0.579, this result is primarily driven by class prevalence rather than meaningful predictive capability. Because the baseline always predicts the majority class, it is incapable of distinguishing between positive and negative cholesterol cases. As a result, the model produced a precision and recall of 0.0 for the positive class and an ROC-AUC score of 0.5, indicating no discriminative ability beyond random chance.

These results highlight a major limitation of naive statistical imputation methods in healthcare-oriented prediction systems. While mode imputation is computationally inexpensive, it fails to leverage relationships between cholesterol status and other health indicators such as BMI, hypertension, age, and cardiovascular history. This motivated the use of an auxiliary machine learning model capable of learning those relationships to produce more informative imputations.

### Auxiliary Cholesterol Prediction Model

#### Model Comparison

Several tabular models were evaluated for the cholesterol prediction model, including Logistic Regression, Random Forest, XGBoost, CatBoost, and LightGBM. The goal for the model was to achieve robust imputation performance while prioritizing ROC-AUC and recall.

````python
from sklearn.ensemble import RandomForestClassifier

# Preprocess data
X = chol_df.drop(columns=["HighChol"])
Y = chol_df["HighChol"]

num_features = X.select_dtypes(exclude=["object"]).columns

numeric_transformer = StandardScaler()

preprocessor = ColumnTransformer(
    [
        ("StandardScaler", numeric_transformer, num_features)
    ]
)

X = preprocessor.fit_transform(X)

# Train test split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)

chol_models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "XGBoost": XGBClassifier(random_state=42, verbosity=0),
    "CatBoost": CatBoostClassifier(random_state=42, verbose=0),
    "LightGBM": LGBMClassifier(random_state=42, verbosity=-1)
}

chol_model_list = []
chol_acc_list = []
chol_prec_list = []
chol_rec_list = []
chol_auc_list = []

for i in range(len(list(chol_models))):

    # Train model
    chol_model = list(chol_models.values())[i]

    chol_model.fit(x_train, y_train)

    # Predict on test set
    y_chol_pred = chol_model.predict(x_test)
    y_chol_proba = chol_model.predict_proba(x_test)[:, 1]

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_chol_pred)
    precision = precision_score(y_test, y_chol_pred)
    recall = recall_score(y_test, y_chol_pred)
    roc_auc = roc_auc_score(y_test, y_chol_proba)

    chol_model_list.append(list(chol_models.keys())[i])
    chol_acc_list.append(accuracy)
    chol_prec_list.append(precision)
    chol_rec_list.append(recall)
    chol_auc_list.append(roc_auc)

    print(list(chol_models.keys())[i])
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"ROC AUC: {roc_auc:.4f}")

    print('=' * 35)
    print("\n")
````

#### Hyperparameter Tuning

Hyperparameter tuning was performed using `RandomizedSearchCV` to improve the performance of the selected CatBoost cholesterol prediction model.

| Parameter | Selected Values |
|-----------|-----------------|
| `depth` | 4, 6, 8 |
| `learning_rate` | 0.01, 0.05, 0.1 |
| `iterations` | 300, 600, 1000 |

The selected search ranges were designed to balance model complexity, generalization performance, and computational efficiency. Moderate tree depths and learning rates were tested to capture nonlinear relationships within the healthcare survey data while reducing overfitting risk. Iteration counts ranging from 300 to 1000 were evaluated to allow sufficient boosting capacity without introducing excessive training time.

Because the evaluated gradient boosting models had already converged to similar baseline performance, the tuning process focused on practical and computationally efficient parameter ranges rather than exhaustive optimization.

````python
from sklearn.model_selection import RandomizedSearchCV

# Get validation sets
x_train_param, x_val, y_train_param, y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42, stratify=y_train)

cat_param_dist = {
    "depth": [4, 6, 8],
    "learning_rate": [0.01, 0.05, 0.1],
    "iterations": [300, 600, 1000]
}

tuning_model = CatBoostClassifier(random_state=42, verbose=0, loss_function="Logloss", auto_class_weights="Balanced")
tuning_model.fit(x_train, y_train)

random_search = RandomizedSearchCV(tuning_model, cat_param_dist, n_iter=20, scoring="roc_auc", cv=3, n_jobs=-1, random_state=42)
random_search.fit(x_train_param, y_train_param, eval_set=[(x_val, y_val)], early_stopping_rounds=10)

accuracy = accuracy_score(y_test, random_search.predict(x_test))
precision = precision_score(y_test, random_search.predict(x_test))
recall = recall_score(y_test, random_search.predict(x_test))
roc_auc = roc_auc_score(y_test, random_search.predict_proba(x_test)[:, 1])

print("Best Hyperparameters:", random_search.best_params_)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("ROC AUC:", roc_auc)
````

The optimal hyperparameters were found to be:
* `learning_rate = 0.05`
* `depth = 6`
* `iterations = 600`

After hyperparameter tuning, recall improved substantially from 0.572 to 0.705, while ROC-AUC increased slightly from 0.733 to 0.735. Precision decreased from 0.630 to 0.592, indicating a shift toward more recall-oriented classification behavior.

Threshold optimization was evaluated but ultimately not implemented because it did not meaningfully improve the precision-recall tradeoff of the tuned model.
